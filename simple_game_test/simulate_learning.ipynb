```python
# Functions to get constraints

domain_map = {
    'at': 'augmented_taxi2',
    'sb': 'skateboard2'
}

def parse_trial_mdp_info(domain, trial_str):
    """
    Parses trial MDP information from a string, extracting environment index,
    optimal moves, and the parsed MDP dictionary.

    Args:
        domain (str): The domain ('at' or 'sb').
        trial_str (str): A string representation of the trial MDP dictionary.

    Returns:
        tuple: (env_idx, opt_moves, trial_mdp_dict). env_idx can be None.
    """
    if domain == 'at':
        mdp_class = 'augmented_taxi2'
    elif domain == 'sb':
        mdp_class = 'skateboard2'
    else:
        raise ValueError(f"Unknown domain: {domain}")

    try:
        trial_mdp_dict = ast.literal_eval(trial_str)
        env_idx = trial_mdp_dict.get('env_traj_idxs', [None])[0]
        opt_moves = trial_mdp_dict.get('opt_actions', [])
        return env_idx, opt_moves, trial_mdp_dict
    except (ValueError, SyntaxError) as e:
        print(f"Warning: Could not parse trial_str '{trial_str}'. Attempting fallback. Error: {e}")
        # Fallback if direct parsing fails, try to find an environment file
        # This part of the original logic seems a bit ambiguous.
        # If the `trial_str` is truly malformed, `ast.literal_eval` will fail again.
        # Re-evaluating `trial_str` within the except block is redundant if it failed once.
        # The original code's commented out section suggests attempting `check_env_files` here,
        # but it was commented out. For clarity, I'll keep the current behavior but note the ambiguity.
        # If `check_env_files` is truly intended as a fallback for missing `env_traj_idxs`,
        # it should be called after successful `ast.literal_eval` but before the return.

        # If trial_str is truly unparseable as a dict, this will likely fail again.
        # Consider a more robust parsing or simply returning defaults if it's consistently malformed.
        try:
            trial_mdp_dict = ast.literal_eval(trial_str)
            opt_moves = trial_mdp_dict.get('opt_actions', [])
            env_idx = check_env_files(mdp_class, trial_mdp_dict) # Call the helper function
            return env_idx, opt_moves, trial_mdp_dict
        except (ValueError, SyntaxError) as e:
            print(f"Failed to parse trial info again: {e}")
            return None, [], {} # Return safe defaults if parsing completely fails


def get_traj_from_mdp(mdp, current_state, moves_list, weights=None):
    """
    Generates a trajectory by executing a sequence of moves in an MDP.

    Args:
        mdp (object): The MDP object.
        current_state (object): The initial state.
        moves_list (list): A list of actions to execute.
        weights (np.ndarray, optional): Reward weights to set on the MDP. Defaults to None.

    Returns:
        list: The generated trajectory, a list of (state, action, next_state) tuples.
    """
    if weights is not None:
        mdp.weights = weights

    trajectory = []
    for move in moves_list:
        # Assumes that the user study only allows actions that change the state of the MDP
        # and that mdp.execute_agent_action updates the internal state of mdp.
        reward, next_state = mdp.execute_agent_action(move)
        trajectory.append((current_state, move, next_state))
        current_state = next_state # Update current_state for the next iteration

    return trajectory


def calculate_incorrect_constraints(domain, mdp_parameters, moves):
    
    sb_gt_weights = np.array([[0.59565914, 0.3519804, -0.72201107]])
    at_gt_weights = np.array([[-0.63599873, 0.74199852, -0.21199958]])

    if domain == 'at':
        gt_weights = at_gt_weights
    elif domain == 'sb':
        gt_weights = sb_gt_weights
    else:
        # Handle cases where domain is not 'at' or 'sb'
        print(f"Warning: Unknown domain '{domain}'. Cannot check reward sign correctness.")
        return False

    # Parse trial MDP info and moves
    env_idx, opt_moves, trial_mdp_dict = parse_trial_mdp_info(domain, mdp_parameters)
    human_moves = ast.literal_eval(moves)
    # human_est_weights = normalize_reward_weights(trial_data['reward_ft_weights']) # Not used in final return, so commented out

    # Make custom MDP (assuming make_custom_mdp handles different classes)
    trial_mdp_dict['weights'] = gt_weights # Set weights for MDP creation
    mdp = make_custom_mdp(domain_map.get(domain), trial_mdp_dict) # Use a clearer domain map

    current_state = mdp.get_init_state()

    # Get optimal and human trajectories under ground truth MDP
    opt_traj = get_traj_from_mdp(mdp, current_state, opt_moves, weights=gt_weights)
    mdp.reset() # Reset MDP state for the next trajectory calculation
    human_traj = get_traj_from_mdp(mdp, current_state, human_moves, weights=gt_weights)
    mdp.reset()

    # Feature counts
    phi_opt = mdp.accumulate_reward_features(opt_traj, discount=True)
    mdp.reset()
    phi_human = mdp.accumulate_reward_features(human_traj, discount=True)
    mdp.reset()
    
    # opt-human is the difference or gap in knowledge, so its inverse is the actual knowledge
    return phi_human - phi_opt
```

```python
def _standardize_id_column(df, column_name):
    """
    Helper function to standardize ID columns: fill NaNs, convert to int then str,
    and replace '-1' with an empty string.

    Args:
        df (pd.DataFrame): The DataFrame to modify.
        column_name (str): The name of the column to standardize.

    Returns:
        pd.DataFrame: The DataFrame with the standardized column.
    """
    # Ensure a copy to avoid SettingWithCopyWarning
    df_copy = df.copy()
    df_copy[column_name] = df_copy[column_name].fillna(-1).astype(int).astype(str)
    df_copy.loc[df_copy[column_name] == '-1', column_name] = ''
    return df_copy

def _flatten_one_more_level(nested_list):
    """Flattens a list of lists one level deeper if sub-elements are also lists."""
    flattened = []
    for item in nested_list:
        if isinstance(item, list) and all(isinstance(sub, list) for sub in item):
            flattened.extend(item)
        else:
            flattened.append(item)
    return flattened


def _is_valid_constraint(c):
    """Checks if a constraint list is valid (not None or empty/all None)."""
    if c is None:
        return False
    # Flatten and check if any element is not None
    flat_list = []
    for item in c:
        if isinstance(item, list):
            for sub_item in item:
                flat_list.append(sub_item)
        else:
            flat_list.append(item)
    return any(x is not None for x in flat_list)
```

## Run learning loop

```python

filename = os.path.abspath('../results/output_full_data_06_25.xlsx')
print(f'Processing file: {filename}')

# Determine file suffix for output naming
file_suffix = filename[len('output_'):] if filename.startswith('output_') else filename
print(f'File suffix: {file_suffix}')

# Read all sheets from the Excel file
excel_file = pd.ExcelFile(filename)
print('Parsing all data from Excel sheets...', flush=True)

# --- Load and Clean DataFrames ---
user_df = excel_file.parse('user')
user_df = _standardize_id_column(user_df, 'group')
user_df = _standardize_id_column(user_df, 'id')

trial_df = excel_file.parse('trial')
trial_df['group_id'] = trial_df['group'].copy() # Create 'group_id' from 'group'
trial_df = _standardize_id_column(trial_df, 'group')
trial_df = _standardize_id_column(trial_df, 'user_id')
trial_df = trial_df.drop_duplicates(subset=['user_id', 'group', 'domain', 'round', 'iteration'])


################

# Group by user
users = trial_df['user_id'].unique()

user_knowledge_df = pd.DataFrame()

print(trial_df['domain'].unique())
print(trial_df['user_id'].unique())

print(len(trial_df))

# Main PF update loop
for user in users:
    
    user_knowledge_dict = {}
    
    for domain in ['at', 'sb']:
        if domain == 'at':
            mdp_class = 'augmented_taxi2'
        elif domain == 'sb':
            mdp_class = 'skateboard2'
        params = get_mdp_parameters(mdp_class)
        params['teacher_update_model_type'] = 'low_noise'
        
        n_particles = 5000
        N_elevation_bins = params['N_elevation_bins']
        step_cost_flag = params['step_cost_flag']
        weights = params['mdp_parameters']['weights']
        teaching_uf = params['teacher_learning_factor']
        model_type = params['teacher_update_model_type']
        
        user_pf = pf_team.Particles_team(sample_human_models_uniform([], n_particles), teaching_uf, N_elevation_bins)
        
        print('Simulating learning for user: ', user, 'domain:', domain)
        
        # user-domain trial data
        user_trial_data = trial_df[(trial_df['user_id']==user) & (trial_df['domain']==domain)]
        
        print('N trials: ', len(user_trial_data))
        
        min_BEC_constraints_running = []

        if len(user_trial_data) > 0:
        
            N_lessons = user_trial_data[user_trial_data['interaction_type']=='diagnostic test'].iloc[-1]['round']

            for lesson in range(1, N_lessons+1):

                demos = user_trial_data[(user_trial_data['interaction_type']=='demo') & (user_trial_data['round']==lesson)]
                tests = user_trial_data[(user_trial_data['interaction_type']=='diagnostic test') & (user_trial_data['round']==lesson)]
                
                print('N demos:', len(demos))
                print('N tests:', len(tests))
                
                demo_constraints = []
                
                for _, demo in demos.iterrows():
                    env_idx, opt_moves, trial_mdp_dict = parse_trial_mdp_info(domain, demo['mdp_parameters'])
                    demo_constraints.append(trial_mdp_dict.get('constraints', []))
                
                demo_constraints_cleaned = [c for c in demo_constraints if _is_valid_constraint(c)]
                demo_constraints_flat = _flatten_one_more_level(demo_constraints_cleaned) if demo_constraints_cleaned else []
                
                test_constraints = []
                feedback_constraints = []
                
                for _, test in tests.iterrows():
                    env_idx, opt_moves, trial_mdp_dict = parse_trial_mdp_info(domain, test['mdp_parameters'])
                    
                    if test['is_opt_response']==1:    
                        test_constraints.append(trial_mdp_dict.get('constraints', []))
                    else:
                        test_constraints.append(calculate_incorrect_constraints(domain, test['mdp_parameters'], test['moves']))
                        feedback_constraints.append(trial_mdp_dict.get('constraints', []))
                        
                test_constraints_cleaned = [c for c in test_constraints if _is_valid_constraint(c)]
                test_constraints_flat = _flatten_one_more_level(test_constraints_cleaned) if test_constraints_cleaned else []
                
                
                if len(demo_constraints_flat) > 0:

                    feedback_constraints_cleaned = [c for c in feedback_constraints if _is_valid_constraint(c)]
                    feedback_constraints_flat = _flatten_one_more_level(feedback_constraints_cleaned) if feedback_constraints_cleaned else []

                    print('demo_constraints_flat: ', demo_constraints_flat)
                    print('test_constraints_flat: ', test_constraints_flat)
                    print('feedback_constraints_flat: ', feedback_constraints_flat)

                    min_demo_constraints = remove_redundant_constraints(demo_constraints_flat, weights, step_cost_flag) # minimum constraints conveyed by the unit's demonstrations
                    min_test_constraints = remove_redundant_constraints(test_constraints_flat, weights, step_cost_flag) # minimum constraints conveyed by the unit's demonstrations

                    if feedback_constraints_flat:
                        min_feedback_constraints = remove_redundant_constraints(feedback_constraints_flat, weights, step_cost_flag)

                    print('min_demo_constraints: ', min_demo_constraints)
                    print('min_test_constraints: ', min_test_constraints)
                    print('min_feedback_constraints: ', min_feedback_constraints)

                    min_BEC_constraints_running.extend(min_demo_constraints)
                    min_BEC_constraints_running = remove_redundant_constraints(min_BEC_constraints_running, weights, step_cost_flag)
                    
                    print('min_BEC_constraints_running: ', min_BEC_constraints_running)
                    
                    # Update the particle filter - demos
                    user_pf.update(min_demo_constraints, teaching_uf, model_type, params)
                    user_pf.calc_particles_probability(min_BEC_constraints_running, true_reward_weights=params['mdp_parameters']['weights'])

                    user_knowledge_dict = {'user_id': user, 'domain': domain, 'round': lesson, 'interaction': 'demo', 'reward_belief': user_pf.particles_prob_correct}
                    user_knowledge_df = user_knowledge_df.append(user_knowledge_dict, ignore_index=True)

                    # Update the particle filter - tests
                    user_pf.update(min_test_constraints, teaching_uf, model_type, params)
                    user_pf.calc_particles_probability(min_BEC_constraints_running, true_reward_weights=params['mdp_parameters']['weights'])

                    user_knowledge_dict = {'user_id': user, 'domain': domain, 'round': lesson, 'interaction': 'diagnostic test', 'reward_belief': user_pf.particles_prob_correct}
                    user_knowledge_df = user_knowledge_df.append(user_knowledge_dict, ignore_index=True)

    break

print(user_knowledge_df)
```
